```python
from pyspark.sql.functions import *

df = spark.range(1,100001) \
.withColumnRenamed("id","order_id") \
.withColumn("store_id", concat(lit("S"), (rand()*20+1).cast("int"))) \
.withColumn("product_id", concat(lit("P"), (rand()*100+1).cast("int"))) \
.withColumn("amount", round(rand()*5000+100,2)) \
.withColumn("quantity", (rand()*5+1).cast("int")) \
.withColumn("order_date", expr("date_add('2024-01-01', cast(rand()*60 as int))"))

display(df)
```

```python
df.write.mode("overwrite").saveAsTable("workspace.default.bronze_sales")
```
